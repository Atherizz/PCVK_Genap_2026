```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
import math
import glob
import os
```

```python
print(' Mengubah tingkat kecerahan citra ')
print('-----------------------------------')
try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')

# sesuaikan dengan file dan folder di drive Anda
original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')
brightness_image = np.zeros(original.shape, original.dtype)

# akses per piksel
# np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            brightness_image[y, x, c] = np.clip(original[y, x, c] + brightness, 0, 255)

# cara simple tanpa for loop
# brightness_image = cv.convertScaleAbs(original, beta=brightness)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
brightness_rgb = cv.cvtColor(brightness_image, cv.COLOR_BGR2RGB)

final_frame = cv.hconcat((original_rgb, brightness_rgb))
plt.figure(figsize=(10, 5))
plt.imshow(final_frame)
plt.axis('off')
plt.show()
```

Tugas Praktikum 1 — Inverse Citra

```python
original = cv.imread('/content/drive/MyDrive/PCVK/pepper.png')
inverse_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            inverse_image[y, x, c] = 255 - original[y, x, c]

# cara simple tanpa for loop
# inverse_image = 255 - original

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
inverse_rgb = cv.cvtColor(inverse_image, cv.COLOR_BGR2RGB)

final_frame = cv.hconcat((original_rgb, inverse_rgb))
plt.figure(figsize=(10, 5))
plt.imshow(final_frame)
plt.axis('off')
plt.title('Original vs Inverse (Negative)')
plt.show()
```

Tugas Praktikum 2 — Transformasi Contrast

```python
print(' Mengubah kontras dan tingkat kecerahan citra ')
print('------------------------------------------------')
try:
    brightness = int(input('Masukkan tingkat kecerahan : '))
    contrast = float(input('Masukkan kontras : '))
except ValueError:
    print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')
contrast_image = np.zeros(original.shape, original.dtype)

factor = (259 * (contrast + 255)) / (255 * (259 - contrast))

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            nilai = factor * (int(original[y, x, c]) - 128) + 128 + brightness
            contrast_image[y, x, c] = np.clip(nilai, 0, 255)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
contrast_rgb = cv.cvtColor(contrast_image, cv.COLOR_BGR2RGB)

final_frame = cv.hconcat((original_rgb, contrast_rgb))
plt.figure(figsize=(10, 5))
plt.imshow(final_frame)
plt.axis('off')
plt.show()
```

Tugas Praktikum 3 — Transformasi Logarithmic Brightness

```python
print(' Mengubah tingkat kecerahan citra dengan Transformasi Log ')
print('-------------------------------------------------------')
try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')
log_image = np.zeros(original.shape, original.dtype)

# konstanta c dihitung supaya hasil transformasi tetap proporsional
# terhadap nilai kecerahan yang diminta user, dan dibatasi ke range 0-255
c = brightness / math.log(1 + 255)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for ch in range(original.shape[2]):
            pixel = int(original[y, x, ch])
            nilai = c * math.log(1 + pixel)
            log_image[y, x, ch] = np.clip(nilai, 0, 255)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
log_rgb = cv.cvtColor(log_image, cv.COLOR_BGR2RGB)

final_frame = cv.hconcat((original_rgb, log_rgb))
plt.figure(figsize=(10, 5))
plt.imshow(final_frame)
plt.axis('off')
plt.show()
```

Tugas Praktikum 4 — Grayscale (Averaging, Lightness, Luminance)

```python
original = cv.imread('/content/drive/MyDrive/PCVK/pepper.png')
original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)

h, w = original.shape[:2]
gray_avg = np.zeros((h, w), dtype=np.uint8)
gray_light = np.zeros((h, w), dtype=np.uint8)
gray_lum = np.zeros((h, w), dtype=np.uint8)

for y in range(h):
    for x in range(w):
        B, G, R = original[y, x]          # OpenCV membaca dalam urutan BGR
        R, G, B = int(R), int(G), int(B)
        gray_avg[y, x]   = (R + G + B) / 3
        gray_light[y, x] = (max(R, G, B) + min(R, G, B)) / 2
        gray_lum[y, x]   = 0.21 * R + 0.72 * G + 0.07 * B

# a. Averaging
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].axis('off')
ax[1].imshow(gray_avg, cmap='gray'); ax[1].axis('off')
plt.show()

# b. Lightness
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].axis('off')
ax[1].imshow(gray_light, cmap='gray'); ax[1].axis('off')
plt.show()

# c. Luminance
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].axis('off')
ax[1].imshow(gray_lum, cmap='gray'); ax[1].axis('off')
plt.show()
```

Tugas Praktikum 5 — Tampilkan Warna Tertentu (Merah), Sisanya Grayscale

```python
original = cv.imread('/content/drive/MyDrive/PCVK/pepper.png')
hsv = cv.cvtColor(original, cv.COLOR_BGR2HSV)

# rentang warna merah pada HSV (merah ada di dua ujung hue, 0 dan 180)
lower_red1 = np.array([0, 70, 50]);   upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 70, 50]); upper_red2 = np.array([180, 255, 255])

mask1 = cv.inRange(hsv, lower_red1, upper_red1)
mask2 = cv.inRange(hsv, lower_red2, upper_red2)
mask_red = cv.bitwise_or(mask1, mask2)

gray = cv.cvtColor(original, cv.COLOR_BGR2GRAY)
gray_bgr = cv.cvtColor(gray, cv.COLOR_GRAY2BGR)

# pixel yang termasuk mask merah -> pakai warna asli, selain itu -> grayscale
hasil = np.where(mask_red[:, :, None] == 255, original, gray_bgr)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
hasil_rgb = cv.cvtColor(hasil, cv.COLOR_BGR2RGB)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].axis('off')
ax[1].imshow(hasil_rgb); ax[1].axis('off')
plt.show()
```

Tugas Praktikum 6 — Gamma Correction

```python
print(' Gamma Correction pada citra ')
print('----------------------------------')
try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')
gamma_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            nilai = 255 * ((original[y, x, c] / 255) ** (1 / gamma))
            gamma_image[y, x, c] = np.clip(nilai, 0, 255)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
gamma_rgb = cv.cvtColor(gamma_image, cv.COLOR_BGR2RGB)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].set_title('Citra Asli'); ax[0].axis('off')
ax[1].imshow(gamma_rgb); ax[1].set_title(f'Gamma Correction (γ = {gamma})'); ax[1].axis('off')
plt.show()
```

Tugas Praktikum 7 — Simulasi Image Depth (Bit Depth)

```python
try:
    bit_depth = int(input('Masukkan bit depth tujuan (1-8): '))
except ValueError:
    print('Error, not a number')

level = 255 / (pow(2, bit_depth) - 1)

original = cv.imread('/content/drive/MyDrive/PCVK/pepper.png', cv.IMREAD_GRAYSCALE)
depth_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        depth_image[y, x] = round((original[y, x] / level)) * level

print(f'Bit depth awal : 8 bit')
print(f'Bit depth tujuan: {bit_depth} bit')
print(f'Jumlah level   : {pow(2, bit_depth)}')

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original, cmap='gray'); ax[0].set_title('Grayscale 8-bit'); ax[0].axis('off')
ax[1].imshow(depth_image, cmap='gray'); ax[1].set_title(f'Grayscale {bit_depth}-bit'); ax[1].axis('off')
plt.show()
```

Tugas Praktikum 8 — Average Denoising

```python
def PSNR(img1, img2):
    mse = np.mean((img1.astype(np.float64) - img2.astype(np.float64)) ** 2)
    if mse == 0:
        return 100
    max_pixel = 255.0
    psnr = 20 * math.log10(max_pixel / math.sqrt(mse))
    return psnr

original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')

cv_img = []
for img in glob.glob('/content/drive/MyDrive/PCVK/noises/*.jpg'):
    n = cv.imread(img)
    cv_img.append(n)

def average_denoise(images, n):
    subset = images[:n]
    stacked = np.stack(subset, axis=0).astype(np.float64)
    avg = np.mean(stacked, axis=0)
    return np.clip(avg, 0, 255).astype(np.uint8)

jumlah_citra_list = [10, 20, 40, 80, 100]
hasil_psnr = {}

for n in jumlah_citra_list:
    denoised = average_denoise(cv_img, n)
    psnr_value = PSNR(original, denoised)
    hasil_psnr[n] = psnr_value
    print(f'Average {n} citra -> PSNR = {psnr_value:.2f} dB')

    denoised_rgb = cv.cvtColor(denoised, cv.COLOR_BGR2RGB)
    plt.figure()
    plt.imshow(denoised_rgb)
    plt.title(f'Average {n} citra (PSNR {psnr_value:.2f} dB)')
    plt.axis('off')
    plt.show()
```

Jawaban pertanyaan:
1. Apakah peningkatan jumlah citra selalu memberikan peningkatan PSNR yang signifikan?
- Tidak. Secara matematis, standar deviasi noise pada averaging turun sebanding dengan 1/√N (N = jumlah citra). Artinya kenaikan PSNR besar terjadi di N kecil (misal dari 10 ke 20 atau 20 ke 40), tetapi begitu N makin besar, penambahan citra memberi penurunan noise yang makin kecil.
2. Pada jumlah citra berapa peningkatan mulai tidak terlalu signifikan?
- Umumnya mulai dari N sekitar 40–50 ke atas, kenaikan PSNR antar tahap (mis. dari 80 ke 100) jauh lebih kecil dibanding kenaikan di tahap awal (10 ke 20).
3. Kesimpulan: Averaging denoising efektif tapi mengalami diminishing returns menambah jumlah citra terus-menerus tidak sepadan dengan biaya komputasi dan penyimpanan tambahan setelah titik tertentu, sehingga perlu dicari titik optimal antara kualitas dan efisiensi.

Tugas Praktikum 9 — Image Masking

```python
from google.colab.patches import cv2_imshow

# ganti path ini dengan foto kamu sendiri jika couple.tiff belum tersedia
original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')

if original is None:
    print("Gambar tidak ditemukan. Cek kembali path filenya.")
else:
    print("Gambar berhasil dibaca.")

    height, width = original.shape[:2]


    # ======================================
    mask = np.zeros((height, width), dtype=np.uint8)
    cv.circle(mask, (int(width * 0.35), int(height * 0.28)),
              int(height * 0.13), 255, -1)
    cv.circle(mask, (int(width * 0.65), int(height * 0.28)),
              int(height * 0.13), 255, -1)
    mask_3ch = cv.cvtColor(mask, cv.COLOR_GRAY2BGR)

    hasil_not  = cv.bitwise_not(mask_3ch)
    hasil_or   = cv.bitwise_or(original, mask_3ch)
    hasil_and  = cv.bitwise_and(original, mask_3ch)
    hasil_nand = cv.bitwise_not(cv.bitwise_and(original, mask_3ch))
    hasil_xor  = cv.bitwise_xor(original, mask_3ch)


    print("\nMASK")
    cv2_imshow(mask)

    print("\n1. NOT (Komplemen)")
    cv2_imshow(hasil_not)

    print("\n2. OR (Atau)")
    cv2_imshow(hasil_or)

    print("\n3. AND (Dan)")
    cv2_imshow(hasil_and)

    print("\n4. NAND (Not And)")
    cv2_imshow(hasil_nand)

    print("\n5. XOR (Exclusive Or)")
    cv2_imshow(hasil_xor)
```

### Tabel Hasil & Analisa

|No.|Operator|Image Input|Image Output|Analisa|
|---|---|---|---|---|
|1|NOT (komplemen)|citra + mask|area lingkaran mask jadi hitam, area luar jadi putih (kebalikan mask asli)|NOT membalik seluruh bit mask (255 → 0, 0 → 255), sehingga area yang tadinya "muncul" kini "tertutup" dan sebaliknya|
|2|OR (Atau)|citra + mask|area lingkaran jadi putih total (255), area luar tetap citra asli|Bitwise OR dengan 255 di area lingkaran memaksa semua bit jadi 1, hasilnya selalu putih penuh di area itu|
|3|AND (Dan)|citra + mask|area lingkaran menampilkan citra asli, area luar jadi hitam|AND dengan 0 di area luar mask memaksa semua bit jadi 0 (hitam), AND dengan 255 mempertahankan nilai asli — inilah prinsip masking standar|
|4|NAND (Not And)|citra + mask|kebalikan dari hasil AND: area lingkaran jadi hitam, area luar tetap citra asli|NAND = NOT(AND), jadi bagian yang tadinya tampil (dalam lingkaran) sekarang disembunyikan, dan sebaliknya|
|5|XOR (Exclusive Or)|citra + mask|area lingkaran mengalami inversi warna (seperti negatif), area luar tidak berubah|XOR dengan 255 membalik semua bit (setara operasi inverse 255-nilai) pada area mask, sedangkan XOR dengan 0 tidak mengubah apa pun di luar mask|

Tugas Praktikum 10 — Foto Malam Hari

```python
original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')
log_image = np.zeros(original.shape, original.dtype)

c = 255 / math.log(1 + 255)  # konstanta agar hasil maksimum tetap 255

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for ch in range(original.shape[2]):
            pixel = int(original[y, x, ch])  # cast dulu, hindari overflow uint8
            nilai = c * math.log(1 + pixel)
            log_image[y, x, ch] = np.clip(nilai, 0, 255)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
log_rgb = cv.cvtColor(log_image, cv.COLOR_BGR2RGB)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].set_title('Sebelum'); ax[0].axis('off')
ax[1].imshow(log_rgb); ax[1].set_title('Sesudah (Log Brightness)'); ax[1].axis('off')
plt.show()
```

Alasan pemilihan: Foto malam hari punya sebagian besar pixel bernilai rendah (gelap) tapi ada detail terang (misal lampu, bintang). Transformasi log memampatkan (compress) rentang nilai terang sambil merentangkan (stretch) rentang nilai gelap, sehingga detail di area gelap jadi lebih terlihat tanpa membuat area terang over-exposed dan terpotong (clipping).

Risiko: Karena log bersifat non-linier, warna hasil bisa terasa "flat"/kurang kontras dibanding aslinya, dan noise pada area gelap ikut terangkat sehingga jadi lebih terlihat (grainy).

Tugas Praktikum 11 — Perbaikan Kualitas Foto crayfish.jpg

```python
original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')

# Tahap 1: Contrast stretching untuk melebarkan histogram yang sempit (foto underwater berkabut)
contrast = 40.0
brightness = 10
factor = (259 * (contrast + 255)) / (255 * (259 - contrast))
step1 = np.zeros(original.shape, original.dtype)
for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            nilai = factor * (int(original[y, x, c]) - 128) + 128 + brightness
            step1[y, x, c] = np.clip(nilai, 0, 255)

# Tahap 2: Gamma correction (<1) untuk mengangkat detail di area gelap
gamma = 0.8
step2 = np.zeros(step1.shape, step1.dtype)
for y in range(step1.shape[0]):
    for x in range(step1.shape[1]):
        for c in range(step1.shape[2]):
            nilai = 255 * ((step1[y, x, c] / 255) ** (1 / gamma))
            step2[y, x, c] = np.clip(nilai, 0, 255)

original_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
hasil_rgb = cv.cvtColor(step2, cv.COLOR_BGR2RGB)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(original_rgb); ax[0].set_title('Before'); ax[0].axis('off')
ax[1].imshow(hasil_rgb); ax[1].set_title('After (Contrast + Gamma)'); ax[1].axis('off')
plt.show()
```

- **Metode yang dipilih:** kombinasi **Transformasi Contrast** lalu **Gamma Correction**.
- **Alasan pemilihan:** foto `crayfish.jpg` diambil di bawah air sehingga kabur/berkabut (low contrast) — histogramnya sempit. Contrast stretching diterapkan dulu untuk melebarkan sebaran intensitas, baru gamma correction (γ < 1) diterapkan untuk mengangkat detail di area gelap tanpa membuat area terang jadi terlalu putih.
- **Parameter terbaik (contoh titik awal, silakan disesuaikan sambil melihat hasil):** `contrast = 40`, `brightness = 10`, `gamma = 0.8`. Coba beberapa nilai contrast (20–60) dan gamma (0.6–1.0) lalu bandingkan mana yang paling menampilkan detail crayfish tanpa membuat warna air jadi tidak natural.
- **Before–After:** ditampilkan oleh `plt.subplots` di atas.
